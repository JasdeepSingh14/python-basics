ANSWER-

6. Explain the concept of generators in Python and how they are defined.

ANSWER- Generators are special iterators in Python that produce values one at a time, on demand (lazy evaluation), instead of storing them all in memory at once.
They are useful when working with large datasets or infinite sequences.
Uses yield instead of return

7. What are the advantages of using generators over regular functions?

ANSWER-
1.Memory Efficiency :– Generators produce values one at a time instead of storing the entire result in memory.
Example: Generating numbers up to 1 billion without storing them all at once.

2.Lazy Evaluation :– Values are computed only when needed, which saves processing time for unused results.
Example: Reading large files line-by-line instead of loading the whole file.

3.Infinite Sequences Support :– Can generate endless data streams without memory overflow.
Example: count() from itertools can generate numbers forever.

4.Improved Performance :– Reduces the overhead of returning large data structures, making loops faster.
Example: Iterating over yield results is faster for large datasets.

5.Cleaner Code :– Removes the need for manual iterator classes; yield automatically maintains state.
Example: A generator can replace a complex __iter__ and __next__ implementation.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python

ANSWER--In Python, map(), filter(), and reduce() are higher-order functions used for processing iterables, but they work differently.

-The map() function applies a given function to each element of an iterable and returns a map object (iterator) containing the transformed elements. For example, map(lambda x: x*2, [1,2]) produces [2,4].

-The filter() function selects elements from an iterable that satisfy a given condition and returns only those matching elements; for instance, filter(lambda x: x%2==0, [1,2,3,4]) produces [2,4].

-The reduce() function, available in the functools module, applies a function cumulatively to the elements of an iterable to reduce it to a single value, such as reduce(lambda x,y: x+y, [1,2,3]) which results in 6.

In short, map() transforms data, filter() extracts specific data, and reduce() combines data into one result.

```python
#1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

def sum_even_numbers(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:  # Check if even
            total += num
    return total

# Example\
nums = [1, 2, 32, 40, 50, 63]
result = sum_even_numbers(nums)
print("Sum of even numbers:", result)
```

```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

# Print all prime numbers from 1 to 200
for n in range(1, 201):
    if is_prime(n):
        print(n, end=" ")
```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms

class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.count = 0
        self.a, self.b = 0, 1  # First two Fibonacci numbers

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.terms:
            if self.count == 0:
                self.count += 1
                return self.a
            elif self.count == 1:
                self.count += 1
                return self.b
            else:
                self.a, self.b = self.b, self.a + self.b
                self.count += 1
                return self.b
        else:
            raise StopIteration


fib = Fibonacci(10)
for num in fib:
    print(num, end=" ")
```

```python
#6. Write a generator function in Python that yields the powers of 2 up to a given exponent

def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

# Example
for value in powers_of_two(5):
    print(value, end=" ")
```

```python
#7. Implement a generator function that reads a file line by line and yields each line as a string.

def read_file_lines(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()
```

```python
#8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple

data = [(12, 34), (3, 2), (24, 68), (52, 41)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)
```

```python
#9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

celsius = [0, 2, 67, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```

```python
#10.  Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(text):
    vowels = "aeiouAEIOU"
    result = filter(lambda ch: ch not in vowels, text)
    return result

# Example usage
text = "Hello World"
filtered_text = remove_vowels(text)
print("Original:", text)
print("Without vowels:", filtered_text)
print(list(filtered_text))
```
